**You do not write code from scratch.** Each cell already holds the code, with blanks

One thing worth knowing before you start: every message in a conversation carries a
**role**, and there are only two. `"user"` is the customer, `"assistant"` is ShopAssist.
That is how Claude tells who said what.

## Setup

Run this cell first, before anything else. Click it, then press **Shift+Enter**.

```python
# --- Lab setup (provided - just run it) ---
import shopassist_lab
from shopassist_lab import check

# Everything below this line is ordinary Claude API code.
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()

model = "claude-sonnet-4-6"

# The two things the customer says, in order.
FIRST_MESSAGE = "I want to return my order."
ORDER_NUMBER_REPLY = "My order number is ORD-12345678."
```

---

## Task 1 - Write the three helper functions

```python
# ============================================================
# TASK 1 - Write the three helper functions
# ============================================================
#
# WHAT TO DO
#   Complete three small functions:
#
#     add_user_message       adds the customer's message to the
#                            conversation list
#     add_assistant_message  adds ShopAssist's reply to the same
#                            list
#     chat                   sends the whole list to Claude and
#                            gives back the reply text
#
# WHY IT MATTERS
#   In the previous lab you typed every message dictionary by
#   hand. That is long, and easy to get wrong. From here on,
#   every notebook in this course uses these three functions
#   instead - so this cell is the one you will reuse the most.
#
#   Notice that chat() does NOT save the reply itself. The
#   caller does that, in the next task. Keeping the two apart is
#   what lets you read, log or change a reply before it becomes
#   part of the history.
#
# WHERE TO SEE IT IN THE LECTURE
#   "First, we need a function that adds a user message" -
#   about 1 minute 4 seconds in.
#
# HOW TO DO IT
#   Three blanks.
#
#   Two of them are roles, and there are only two to choose
#   from:
#
#       "user"        the customer speaking
#       "assistant"   ShopAssist speaking
#
#   The third blank is the conversation list that was handed to
#   chat() - it has a name, right there in the def line.
# ============================================================

# BEGIN SOLUTION
def add_user_message(messages, text):
    messages.append({"role": "user", "content": text})


def add_assistant_message(messages, text):
    messages.append({"role": "assistant", "content": text})


def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=300,
        messages=messages,
    )
    return message.content[0].text
# SCAFFOLD: def add_user_message(messages, text):
# SCAFFOLD:     messages.append({"role": ..., "content": text})   # "user" or "assistant"? this message came from the customer
# SCAFFOLD:
# SCAFFOLD:
# SCAFFOLD: def add_assistant_message(messages, text):
# SCAFFOLD:     messages.append({"role": ..., "content": text})   # "user" or "assistant"? this message came from ShopAssist
# SCAFFOLD:
# SCAFFOLD:
# SCAFFOLD: def chat(messages):
# SCAFFOLD:     message = client.messages.create(
# SCAFFOLD:         model=model,
# SCAFFOLD:         max_tokens=300,
# SCAFFOLD:         messages=...,   # the list this function received - see the def line above
# SCAFFOLD:     )
# SCAFFOLD:     # content[0].text is the reply text, the same as in the earlier labs
# SCAFFOLD:     return message.content[0].text
# END SOLUTION: replace each ... below with the value named beside it

check("helpers", add_user_message=add_user_message,
      add_assistant_message=add_assistant_message, chat=chat)
```

---

## Task 2 - Run the first exchange through them

```python
# ============================================================
# TASK 2 - Run the first exchange through them
# ============================================================
#
# WHAT TO DO
#   Use the three helpers to do one full exchange: the customer
#   speaks, Claude answers, and the answer goes back into the
#   list.
#
# WHY IT MATTERS
#   That last step is the one people forget. If Claude's reply
#   is not saved, the next request will not contain it - and the
#   conversation breaks exactly the way it broke in the previous
#   lab.
#
#   The messages list is a plain Python list living in your
#   program. It is the entire memory of this conversation.
#
# WHERE TO SEE IT IN THE LECTURE
#   "Now let's use these helper functions together" - about
#   3 minutes 27 seconds in.
#
# HOW TO DO IT
#   Three blanks, and each one is a name that already exists in
#   this notebook:
#
#       FIRST_MESSAGE   what the customer says first
#       messages        the conversation list
#       answer          the text chat() gave back
# ============================================================

messages = []

# BEGIN SOLUTION
add_user_message(messages, FIRST_MESSAGE)

answer = chat(messages)

add_assistant_message(messages, answer)
# SCAFFOLD: add_user_message(messages, ...)   # the FIRST_MESSAGE variable
# SCAFFOLD:
# SCAFFOLD: answer = chat(...)   # the conversation list to send - it is called messages
# SCAFFOLD:
# SCAFFOLD: add_assistant_message(messages, ...)   # the answer variable, so the reply is remembered
# END SOLUTION: replace each ... below with the value named beside it

print("Customer:   ", FIRST_MESSAGE)
print()
print("ShopAssist: ", answer)
print()
print("The conversation list now holds", len(messages), "turns:")
for turn in messages:
    print("  {:>9} | {}".format(turn["role"], turn["content"][:60].replace("\n", " ")))

check("first_exchange", messages=messages)
```

---

## Task 3 - Add the follow-up to the same list

```python
# ============================================================
# TASK 3 - Add the follow-up to the same list
# ============================================================
#
# WHAT TO DO
#   The customer sends a second message. Add it to the SAME
#   messages list, send it, and save the reply again.
#
# WHY IT MATTERS
#   There is no messages = [] in this cell, and that is the
#   whole point. You are not starting a new conversation - you
#   are adding to the one you already have. Because the list
#   still holds the first exchange, Claude receives the full
#   context and can answer properly.
#
#   Claude did not remember anything. Your application did.
#
# WHERE TO SEE IT IN THE LECTURE
#   "Now let's add the customer's follow-up message" - about
#   4 minutes 26 seconds in.
#
# HOW TO DO IT
#   Three blanks, the same three names as in Task 2 - except
#   the customer is saying something new this time:
#
#       ORDER_NUMBER_REPLY   the customer's second message
#       messages             the same list, not a new one
#       answer               the text chat() gave back
# ============================================================

# BEGIN SOLUTION
add_user_message(messages, ORDER_NUMBER_REPLY)

answer = chat(messages)

add_assistant_message(messages, answer)
# SCAFFOLD: add_user_message(messages, ...)   # the ORDER_NUMBER_REPLY variable
# SCAFFOLD:
# SCAFFOLD: answer = chat(...)   # the same conversation list, called messages
# SCAFFOLD:
# SCAFFOLD: add_assistant_message(messages, ...)   # the answer variable again
# END SOLUTION: replace each ... below with the value named beside it

print("Customer:   ", ORDER_NUMBER_REPLY)
print()
print("ShopAssist: ", answer)
print()
print("The conversation list now holds", len(messages), "turns:")
for turn in messages:
    print("  {:>9} | {}".format(turn["role"], turn["content"][:60].replace("\n", " ")))

check("two_turns", messages=messages)
```
