> **Update (September 2026) - `temperature` is being retired**
>
> This lesson still works exactly as recorded **if** you installed the SDK with the pinned
> line from notebook 01 (`pip install "anthropic<1"`) and keep `model = "claude-sonnet-4-6"`.
> Two things have changed since the recording:
>
> 1. **SDK.** `anthropic` 1.0+ no longer accepts `temperature=` in `client.messages.create()`
>    and raises `TypeError: ... unexpected keyword argument 'temperature'` before the request
>    is sent. On the 1.x SDK the only way to still send it is `extra_body={"temperature": 0}`.
> 2. **Models.** Claude Sonnet 5, Opus 4.7 and newer reject the parameter at the API level
>    (HTTP 400), so it is on its way out entirely. Prompt instructions (and, on newer models,
>    `output_config={"effort": ...}`) are the recommended way to steer consistency now.
>
> The idea taught here - low temperature for repeatable, policy-driven replies, higher for
> brainstorming - is still correct. Just know that the knob itself is disappearing, and that
> `temperature=0` never guaranteed byte-identical output even on older models.

```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()

model = "claude-sonnet-4-6"
```

```python
message = client.messages.create(
    model=model,
    max_tokens=300,
    temperature=0,  # requires anthropic<1 and a pre-Sonnet-5 model; on SDK 1.x use extra_body={"temperature": 0}
    stop_sequences=["<END>"],
    system="You are ShopAssist AI, a helpful customer support assistant.",
    messages=[
        {
            "role": "user",
            "content": """
Write a short customer support reply about a return request.
End the reply with <END>.
            """
        }
    ]
)

print(message.content[0].text)
print(message.stop_reason)
print(message.stop_sequence)
```
